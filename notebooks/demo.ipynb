# Simulate closed-loop
After implementing your control functionality, you can simulate the closed-loop with code that looks something like this...

```python
#importing from relevant modules
from uuv_mission.dynamic import Submarine, Mission, ClosedLoop
from uuv_mission.control import PDController

sub = Submarine()
# Controller Implementation
controller = PDController()
closed_loop = ClosedLoop(sub, controller)
mission = Mission.from_csv("../data/mission.csv") 

###Checking if the code works
###print(len(mission.reference), mission.reference[0], mission.cave_height[0], mission.cave_depth[0])


trajectory = closed_loop.simulate_with_random_disturbances(mission)
trajectory.plot_completed_mission(mission)
```
